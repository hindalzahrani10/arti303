# Part 1 · Lists

A **list** holds many values in order, in a single variable. Write it
with square brackets, values separated by commas.

````python
names = ["Sara", "Faisal", "Lama"]
gpas = [4.60, 3.90, 4.85]

print(names)
print(gpas)
print("Number of students:", len(names))
````

`len()` gives the number of items. Three students, three names —
and it would still be one variable if there were three hundred.

### Getting items out — indexing

Positions start at **0**, not 1. This trips up almost everyone at first.

````python
print(names[0])     # first item
print(names[1])     # second item
print(names[-1])    # last item  (negative counts from the end)
print(names[-2])    # second from last
````

### Slicing — getting a range of items

`list[start:stop]` gives you items from `start` up to **but not
including** `stop`.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:3])    # positions 1 and 2  ->  [20, 30]
print(numbers[:3])     # from the start     ->  [10, 20, 30]
print(numbers[2:])     # to the end         ->  [30, 40, 50]
````

> **Why "up to but not including"?** It means `numbers[:3]` and
> `numbers[3:]` split the list perfectly with no overlap and nothing
> missed. You'll rely on this constantly when splitting data into
> training and test sets later in the course.

### Changing a list

Lists are **mutable** — you can change them after creating them.

````python
fruits = ["apple", "banana"]

fruits.append("cherry")        # add one item to the end
print(fruits)

fruits.insert(0, "apricot")    # insert at a position
print(fruits)

fruits.remove("banana")        # remove by value
print(fruits)

last = fruits.pop()            # remove and return the last item
print(last, "removed ->", fruits)

fruits[0] = "avocado"          # replace by position
print(fruits)
````

### Useful list operations

````python
scores = [88, 62, 95, 71]

print("sorted:      ", sorted(scores))              # returns a NEW sorted list
print("original:    ", scores)                      # unchanged
print("sorted desc: ", sorted(scores, reverse=True))
print("max, min:    ", max(scores), min(scores))
print("sum, average:", sum(scores), sum(scores) / len(scores))
print("is 95 in it? ", 95 in scores)                # membership test
````

> `sorted(scores)` returns a **new** list and leaves the original
> alone. `scores.sort()` changes the original in place and returns
> `None`. Mixing these up is one of the most common beginner bugs —
> if a sorted list comes out as `None`, this is why.

# Part 2 · Looping over a list

This is where lists become powerful. A `for` loop runs the same code
once for each item.

````python
names = ["Sara", "Faisal", "Lama"]

for name in names:
    print("Hello,", name)
````

`name` is just a variable that takes each value in turn. The
indented block runs once per item.

### Building a new list with a loop

A very common pattern: start with an empty list, append as you go.

````python
gpas = [4.60, 3.90, 4.85, 2.10]
deans_list = []

for gpa in gpas:
    if gpa >= 4.50:
        deans_list.append(gpa)

print("Dean's List GPAs:", deans_list)
````

### Looping with the position too

When you need the index as well as the value, use `enumerate()`.

````python
names = ["Sara", "Faisal", "Lama"]

for i, name in enumerate(names):
    print(f"{i}: {name}")
````

### Looping over two lists together

`zip()` pairs up items from two lists.

````python
names = ["Sara", "Faisal", "Lama"]
gpas = [4.60, 3.90, 4.85]

for name, gpa in zip(names, gpas):
    print(f"{name}: {gpa}")
````

# Part 3 · Tuples

A **tuple** is like a list, but it **cannot be changed** after it's
created. Written with round brackets instead of square ones.

````python
point = (3, 7)
rgb_red = (255, 0, 0)

print(point)
print("x =", point[0], " y =", point[1])   # indexing works the same as a list
print("length:", len(point))
````

````python
# Trying to change a tuple raises an error — this is the whole point of tuples
try:
    point[0] = 99
except TypeError as e:
    print("Error:", e)
````

### Why would you want something you can't change?

Because sometimes "can't change" is exactly the guarantee you need:

- An **image size** `(1920, 1080)` — width and height belong together and
  shouldn't be accidentally modified
- **RGB colour** `(255, 0, 0)` — always exactly three values
- A **database row** — fixed set of fields

Rule of thumb: use a **list** for a collection of similar things that
might grow or shrink. Use a **tuple** for a fixed group of related
values that belong together.

### Unpacking

Both tuples and lists can be "unpacked" into separate variables in one
line — you already did this in Lab 1 without knowing the name.

````python
dimensions = (1920, 1080)
width, height = dimensions

print("width:", width)
print("height:", height)

# A function can return several values — it's really returning a tuple
def min_and_max(numbers):
    return min(numbers), max(numbers)

low, high = min_and_max([4, 9, 1, 7])
print("low:", low, " high:", high)
````

# Part 4 · Sets

A **set** holds items with **no duplicates** and **no order**. Written
with curly braces.

````python
tags = {"ai", "python", "ai", "data", "python"}
print(tags)                  # duplicates disappear automatically
print("count:", len(tags))
````

> The printed order may look scrambled, and may differ between runs.
> That's expected — a set has no order. You cannot do `tags[0]`.

### The most common use: removing duplicates

````python
student_ids = [101, 102, 101, 103, 102, 104]

unique_ids = set(student_ids)
print("unique:", unique_ids)
print("how many unique:", len(unique_ids))

# convert back to a list if you need indexing or order
unique_list = sorted(set(student_ids))
print("as a sorted list:", unique_list)
````

### Comparing two sets

This is where sets get genuinely useful.

````python
week1 = {"Sara", "Faisal", "Lama"}
week2 = {"Faisal", "Lama", "Omar"}

print("attended both:      ", week1 & week2)   # intersection
print("attended either:    ", week1 | week2)   # union
print("only week 1:        ", week1 - week2)   # difference
print("not in both:        ", week1 ^ week2)   # symmetric difference
````

# Part 5 · Dictionaries

A **dictionary** stores **key → value** pairs. Instead of looking things
up by position, you look them up by name.

This finally solves the problem we opened with.

````python
student = {
    "name": "Sara",
    "age": 20,
    "gpa": 4.60,
    "enrolled": True,
}

print(student)
print(student["name"])      # look up by key, not position
print(student["gpa"])
````

Compare that to Lab 1's `name1, age1, gpa1` — one variable now holds
one complete student, and each piece is labelled.

### Adding, changing, and safe lookup

````python
student["section"] = "01"          # add a new key
student["gpa"] = 4.75             # change an existing one

print(student)

# Looking up a key that doesn't exist raises an error:
try:
    print(student["email"])
except KeyError as e:
    print("KeyError:", e)

# .get() returns None (or a default) instead of crashing
print(student.get("email"))
print(student.get("email", "not provided"))
````

> Use `.get()` whenever a key might legitimately be missing — it's the
> difference between a program that handles messy data and one that
> crashes on the first gap. Real datasets are full of gaps.

### Looping over a dictionary

````python
student = {"name": "Sara", "age": 20, "gpa": 4.60}

print("keys:  ", list(student.keys()))
print("values:", list(student.values()))
print()

for key, value in student.items():
    print(f"{key}: {value}")
````

### A list of dictionaries — the shape of real data

This combination is everywhere. It's essentially what a spreadsheet or
a CSV file looks like in Python, and it's exactly what you'll load into
Pandas in Week 6.

````python
students = [
    {"name": "Sara",   "gpa": 4.60, "enrolled": True},
    {"name": "Faisal", "gpa": 3.90, "enrolled": True},
    {"name": "Lama",   "gpa": 4.85, "enrolled": False},
]

for s in students:
    print(f"{s['name']}: {s['gpa']}")

print()
gpa_values = [s["gpa"] for s in students]       # a "list comprehension" — a compact loop
print("average GPA:", sum(gpa_values) / len(gpa_values))
````

> That last line is a **list comprehension** — a shorter way to write
> "build a list by looping." `[s["gpa"] for s in students]` does the
> same thing as creating an empty list and appending in a `for` loop.
> You'll see it constantly in Python code. Use whichever you find
> clearer for now; both are correct.

# Part 6 · Choosing the right one

| | Ordered? | Changeable? | Duplicates? | Written with | Use it when |
|---|---|---|---|---|---|
| **List** | Yes | Yes | Yes | `[ ]` | A collection that may grow, shrink, or need sorting |
| **Tuple** | Yes | **No** | Yes | `( )` | A fixed group of related values that belong together |
| **Set** | No | Yes | **No** | `{ }` | You need uniqueness, or to compare two groups |
| **Dict** | Yes* | Yes | Keys unique | `{k: v}` | Each value needs a label, not just a position |

\* Dictionaries remember insertion order in modern Python, but you
normally access them by key rather than position anyway.

# Part 7 · Functions over collections

A **function** packages code so you can reuse it. Combined with the
structures above, this is how real data code gets written.

````python
def average(numbers):
    """Return the average of a list of numbers. Returns 0.0 if empty."""
    if not numbers:
        return 0.0
    return sum(numbers) / len(numbers)


print(average([4.60, 3.90, 4.85]))
print(average([]))          # the empty case doesn't crash
````

> `if not numbers:` is the Pythonic way to check "is this empty?" —
> it's true for an empty list, empty string, or empty dict. Guarding
> against the empty case is a habit worth building now; a lot of
> real-world crashes are exactly this.

### Exercise 7.2 — putting it all together
Write `summarise(roster)` that takes a list of student dictionaries and

````python
def summarise(roster):
    """Return a summary dictionary for a list of student records."""
    # TODO: your code here
    if not roster:
        return {"count": 0, "average_gpa": 0.0, "dean_list": []}

    gpas = []
    dean_list = []

    for student in roster:
        gpas.append(student["gpa"])
        if student["gpa"] >= 4.50:
            dean_list.append(student["name"])

    return {
        "count": len(roster),
        "average_gpa": sum(gpas) / len(gpas),
        "dean_list": dean_list
    }


sample = [
    {"name": "Sara",   "gpa": 4.60},
    {"name": "Faisal", "gpa": 3.90},
    {"name": "Lama",   "gpa": 4.85},
]

result = summarise(sample)
print(result)
````

# Part 8 · Creating reusable modules

Everything so far has lived inside this notebook. That's fine while
you're exploring — but a notebook is a bad place to *keep* code you
want to reuse.

If `summarise()` only exists in this notebook, then next week, when you
need it again, your options are to copy-paste it or to re-open this
file and hunt for the cell. Neither scales.

The solution is a **module**: a plain `.py` file containing functions,
which any notebook or script can then **import**.

### Writing a module from a notebook

The `%%writefile` command at the top of a cell saves that cell's
contents to a file instead of running it. Run the cell below — it
creates a real file called `student_utils.py` next to this notebook.

````python
%%writefile student_utils.py
"""Reusable helpers for working with student records (GPA out of 5.00)."""

DEANS_LIST_MIN = 4.50


def average_gpa(roster):
    """Return the average GPA of a roster. Returns 0.0 if empty."""
    if not roster:
        return 0.0
    gpas = [student["gpa"] for student in roster]
    return sum(gpas) / len(gpas)


def dean_list_names(roster):
    """Return the names of students at or above the Dean's List threshold."""
    names = []
    for student in roster:
        if student["gpa"] >= DEANS_LIST_MIN:
            names.append(student["name"])
    return names
````

Notice what went into that file:

- A **docstring** at the top saying what the module is for
- `DEANS_LIST_MIN = 4.50` as a **named constant** instead of the number
  `4.50` scattered through the code. If the university ever changes the
  threshold, you edit one line — not six.
- Functions with docstrings, and no `print()` calls. A module
  **returns** values; whatever imports it decides what to display.

### Importing and using it

````python
import student_utils

roster = [
    {"name": "Sara",   "gpa": 4.60},
    {"name": "Faisal", "gpa": 3.90},
    {"name": "Lama",   "gpa": 4.85},
]

print("average:   ", student_utils.average_gpa(roster))
print("dean list: ", student_utils.dean_list_names(roster))
print("threshold: ", student_utils.DEANS_LIST_MIN)
````

You can also import specific names directly, which saves typing the
module name every time:

````python
from student_utils import average_gpa, DEANS_LIST_MIN

print(average_gpa(roster), "vs threshold", DEANS_LIST_MIN)
````

> **The one gotcha to remember:** Python loads a module **once** per
> session. If you edit `student_utils.py` after importing it, the
> notebook will keep using the old version until you **Restart Kernel
> and Run All**. When a change to a `.py` file "does nothing," this is
> almost always why — it will happen to you at least once this
> semester.

A set is unordered so converting the list to a set can change the original order. This could be a problem if the order represents something important, such as student rankings.

**Q3.** Write your own version below that **preserves** the original
order, so `[3, 1, 2, 3, 1]` gives `[3, 1, 2]`. Then say which version
you'd rather have in a real project, and why.

I would prefer the order preserving version because the original order may be important in a real project.

````python
def remove_duplicates_ordered(items):
    """Remove duplicates while keeping the original order."""
    # TODO: your version here
    result = []

    for item in items:
        if item not in result:
            result.append(item)

    return result


print(remove_duplicates_ordered([3, 1, 2, 3, 1]))
````
